```python
from dataclasses import dataclass
from pydantic_ai import Agent, RunContext
from dotenv import load_dotenv
import os

# Carregar variáveis de ambiente
load_dotenv()

#Chamando o agente
agent = Agent(
    'groq:openai/gpt-oss-120b',
    system_prompt="Seja conciso, responda com uma única frase."
)
```

```python
# Parte 1: Execução síncrona convertida para assíncrona
async def consulta_ceara():
    result = await agent.run("Qual é a capital do Ceará?")
    print(result.output)
    return result

# Parte 2: Execução via streaming (sem asyncio.run)
async def consulta_brasil_stream():
    async with agent.run_stream("Qual é a capital da Brasil?") as stream:
        print(await stream.get_output())
    return "Streaming concluído"

# Execute as funções sequencialmente
async def executar_tudo():
    # Primeiro executa a consulta sobre a Itália
    await consulta_ceara()
    
    print("\n--- Agora via streaming ---\n")
    
    # Depois executa a consulta sobre a França com streaming
    await consulta_brasil_stream()

# Basta executar esta linha no Jupyter
await executar_tudo()
```
