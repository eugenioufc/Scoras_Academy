```python
import os
from dotenv import load_dotenv

# Carrega as variáveis do arquivo .env
load_dotenv()

# Agora a variável está disponível como uma variável de ambiente normal
from pydantic_ai import Agent

agent = Agent('openai:gpt-5-nano', system_prompt="Seja conciso...")

async def fazer_pergunta():
    result = await agent.run("De onde vem o 'hello world'?")
    print(result.output)
    return result

await fazer_pergunta()
```

```python
#Exemplo de prompt dinâmico

from pydantic_ai import Agent, RunContext

agent = Agent('openai:gpt-5-nano', deps_type=str, system_prompt="Use o nome do usuário na resposta.")

@agent.system_prompt
async def adicionar_nome(ctx: RunContext[str]) -> str:
    return f"O nome do usuário é {ctx.deps}."

# Função assíncrona para executar o agente
async def executar_consulta():
    result = await agent.run("Qual é a data de hoje?", deps="Eugenio JR")
    print(result.output)
    return result

# Chamada assíncrona
await executar_consulta()
```

```python

from pydantic_ai import Agent
from pydantic import BaseModel, Field
from pydantic_ai import Agent
class ChatResponse(BaseModel):
    message: str = Field(..., description="Mensagem principal")
    timestamp: str = Field(..., description="Horário da resposta")

agent = Agent('openai:gpt-5-nano', output_type=ChatResponse, system_prompt="Responda de forma detalhada citando a data e hora da resposta.")

async def executar_chat():
    result = await agent.run("Olá, tudo bem?")
    print(result.output)
    return result

await executar_chat()
```
